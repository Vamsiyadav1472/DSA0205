```python
import cv2
import matplotlib.pyplot as plt

from google.colab import files
uploaded = files.upload()

image = cv2.imread('/content/IMG_20230723_095508.jpg')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

output = image.copy()

for cnt in contours:
    epsilon = 0.02 * cv2.arcLength(cnt, True)
    approx = cv2.approxPolyDP(cnt, epsilon, True)
    cv2.drawContours(output, [approx], -1, (0,255,0), 2)

plt.imshow(cv2.cvtColor(output, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
# ============================================================
# Contours (Hierarchy, Area, Perimeter, Orientation, Convexity)
# ============================================================

# Install OpenCV (Run once in Google Colab)
!pip install opencv-python -q

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# -------------------------
# Upload Image
# -------------------------
uploaded = files.upload()

# Replace with your uploaded image name
image = cv2.imread("/content/IMG_20230723_095508.jpg")

# Check image loaded
if image is None:
    raise Exception("Image not found. Please check the filename.")

# Convert to RGB for displaying
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Convert to Grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Binary Threshold
_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# -------------------------
# Find Contours
# -------------------------
contours, hierarchy = cv2.findContours(
    thresh,
    cv2.RETR_TREE,
    cv2.CHAIN_APPROX_SIMPLE
)

print("="*60)
print("CONTOUR INFORMATION")
print("="*60)

print("\nHierarchy:")
print(hierarchy)

output = image_rgb.copy()

# -------------------------
# Process Every Contour
# -------------------------
for i, cnt in enumerate(contours):

    print(f"\nContour {i+1}")
    print("-"*40)

    # Area
    area = cv2.contourArea(cnt)
    print("Area :", area)

    # Perimeter
    perimeter = cv2.arcLength(cnt, True)
    print("Perimeter :", perimeter)

    # Convexity
    convex = cv2.isContourConvex(cnt)
    print("Convex :", convex)

    # Draw contour
    cv2.drawContours(output, [cnt], -1, (0,255,0), 2)

    # Draw Convex Hull
    hull = cv2.convexHull(cnt)
    cv2.drawContours(output, [hull], -1, (255,0,0), 2)

    # Orientation
    if len(cnt) >= 5:
        ellipse = cv2.fitEllipse(cnt)
        angle = ellipse[2]
        print("Orientation (degrees):", round(angle,2))

        center = (int(ellipse[0][0]), int(ellipse[0][1]))
        cv2.putText(output,
                    f"{round(angle,1)}",
                    center,
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.5,
                    (255,0,255),
                    2)

# -------------------------
# Display Result
# -------------------------
plt.figure(figsize=(10,8))
plt.imshow(output)
plt.title("Contours: Hierarchy | Area | Perimeter | Orientation | Convexity")
plt.axis("off")
plt.show()
```

```python
# ============================================================
# Contour Detection Using Multiple Edge Detection Techniques
# (Canny, Sobel, LoG, Scharr, Deep Learning HED)
# ============================================================

# Install OpenCV
!pip install opencv-python -q

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# ----------------------------
# Upload Image
# ----------------------------
uploaded = files.upload()

# Change filename if needed
image = cv2.imread("/content/IMG_20230723_095508.jpg")

if image is None:
    raise Exception("Image not found!")

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# ============================================================
# 1. Canny Edge Detection
# ============================================================

canny = cv2.Canny(gray,100,200)

# ============================================================
# 2. Sobel Operator
# ============================================================

sobelx = cv2.Sobel(gray,cv2.CV_64F,1,0,ksize=3)
sobely = cv2.Sobel(gray,cv2.CV_64F,0,1,ksize=3)

sobel = cv2.magnitude(sobelx,sobely)
sobel = np.uint8(sobel)

# ============================================================
# 3. Laplacian of Gaussian (LoG)
# ============================================================

blur = cv2.GaussianBlur(gray,(3,3),0)
log = cv2.Laplacian(blur,cv2.CV_64F)
log = np.uint8(np.absolute(log))

# ============================================================
# 4. Scharr Operator
# ============================================================

scharrx = cv2.Scharr(gray,cv2.CV_64F,1,0)
scharry = cv2.Scharr(gray,cv2.CV_64F,0,1)

scharr = cv2.magnitude(scharrx,scharry)
scharr = np.uint8(scharr)

# ============================================================
# 5. Deep Learning Based Edge Detection (HED)
# ============================================================

try:

    print("Downloading HED model...")

    !wget -q https://raw.githubusercontent.com/opencv/opencv_extra/master/testdata/dnn/hed_deploy.prototxt -O deploy.prototxt

    !wget -q https://github.com/s9xie/hed/releases/download/v1.0/hed_pretrained_bsds.caffemodel

    net = cv2.dnn.readNetFromCaffe(
        "deploy.prototxt",
        "hed_pretrained_bsds.caffemodel"
    )

    H,W = image.shape[:2]

    blob = cv2.dnn.blobFromImage(
        image,
        scalefactor=1.0,
        size=(W,H),
        mean=(104.00698793,116.66876762,122.67891434),
        swapRB=False,
        crop=False
    )

    net.setInput(blob)

    hed = net.forward()
    hed = hed[0,0]
    hed = cv2.resize(hed,(W,H))

except:

    print("HED model could not be loaded.")
    hed = np.zeros_like(gray)

# ============================================================
# Display Results
# ============================================================

plt.figure(figsize=(18,10))

plt.subplot(2,3,1)
plt.imshow(cv2.cvtColor(image,cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(2,3,2)
plt.imshow(canny,cmap="gray")
plt.title("Canny Edge Detection")
plt.axis("off")

plt.subplot(2,3,3)
plt.imshow(sobel,cmap="gray")
plt.title("Sobel Operator")
plt.axis("off")

plt.subplot(2,3,4)
plt.imshow(log,cmap="gray")
plt.title("Laplacian of Gaussian")
plt.axis("off")

plt.subplot(2,3,5)
plt.imshow(scharr,cmap="gray")
plt.title("Scharr Operator")
plt.axis("off")

plt.subplot(2,3,6)
plt.imshow(hed,cmap="gray")
plt.title("Deep Learning (HED)")
plt.axis("off")

plt.tight_layout()
plt.show()
```
