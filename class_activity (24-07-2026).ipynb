```
import cv2
import matplotlib.pyplot as plt

image = cv2.imread('/content/IMG_20231012_081320.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

mean = cv2.blur(image, (5,5))

plt.imshow(mean)
plt.title("Mean Filter")
plt.axis("off")
plt.show()
```

```
import cv2
import matplotlib.pyplot as plt

image = cv2.imread('/content/IMG_20231012_081320.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

median = cv2.medianBlur(image, 5)

plt.imshow(median)
plt.title("Median Filter")
plt.axis("off")
plt.show()
```

```
import cv2
import matplotlib.pyplot as plt

image = cv2.imread('/content/IMG_20231012_081320.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

gaussian = cv2.GaussianBlur(image, (5,5), 0)

plt.imshow(gaussian)
plt.title("Gaussian Filter")
plt.axis("off")
plt.show()
```

```
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('/content/IMG_20231012_081320.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

kernel = np.array([[0,-1,0],
                   [-1,5,-1],
                   [0,-1,0]])

spatial = cv2.filter2D(image, -1, kernel)

plt.imshow(spatial)
plt.title("Spatial Filtering")
plt.axis("off")
plt.show()
```

```
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('/content/IMG_20231012_081320.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

kernel = np.array([[0,-1,0],
                   [-1,5,-1],
                   [0,-1,0]])

sharp = cv2.filter2D(image, -1, kernel)

plt.imshow(sharp)
plt.title("Image Sharpening")
plt.axis("off")
plt.show()
```
